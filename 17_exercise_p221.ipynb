```python
# 인덱스 행으로 컬럼시작하기(쌤)

covid = pd.read_csv('covid.csv', index_col=0, encoding='cp949')
covid
```

```python
# ex)2-2 사례수가 가장 높은 항목의 모든 정보를 출력하기

covid1 = covid.sort_values(by='사례수', ascending= False)
covid1.head(1)
```

```python
# 사례수가 가장 높은 것(쌤)
covid['사례수'].max()
```

```python
# 스마트폰이 가장 높다 -> 그 행 다 추출(쌤)

covid[covid['사례수'] == covid['사례수'].max()]
```

```python
# ex)2-3 증가율이 높은 3개의 항목 출력하기
 
covid3 = covid.sort_values(by='증가', ascending= False)
covid3.head(3)
```

```python
# 증가율이 높은 3개 --> 정렬 --> 내림차순(쌤)

covid3 = covid.sort_values(by='증가', ascending= False)
covid3.head(3)
```

```python
# ex)3

vaccine = pd.read_csv('vaccine.csv', encoding='cp949')

vaccine
```

```python
# ex) 3-1  하위 5행의 데이터를 확인하기

vaccine.tail(5)
```

```python
# ex) 3-2 필요없는 열 삭제하기
# axis=1  열(컬럼)기준으로 지우기. 미지기입시 행이 지워짐.

vaccine1 = vaccine.drop(['day1sum', 'day1rate', 'day2sum', 'day2rate'], axis=1)
vaccine1
```

```python
# ex) 3-3 기초통계량 확인하기
# describe()

vaccine1.describe()
```

```python
# ex) 3-4 월별 1차접종사, 2차접종자수 컬럼의 평균을 출력하세요.


vaccine1['date'] = pd.to_datetime(vaccine1['date'] )
vaccine1.info()
```

```python
vaccine1['month'] = vaccine1['date'].dt.month
vaccine1
```

```python
# 월별 day1num, day2num

vaccine2 = vaccine1[['day1num', 'day2num', 'month']]
vaccine2.head()
```

```python
# 월별 day1num, day2num

vaccine2.groupby('month').mean()
```

```python
# ex) 4

df1 =pd.read_csv('signup.csv', encoding='cp949')
df1
```

```python
df2 =pd.read_csv('professor.csv', encoding='cp949')
df2
```

```python
# ex) merge()

df = pd.merge(df1, df2, on='담당교수')
df
```

```python
# 교수테이블에서 학과별 인원수 출력하기
# 학과별 인원수 --> 그룹 --> count()
df3 = df[['담당교수', '학과']]
df3 = df3.groupby('학과').count()
df3
```
