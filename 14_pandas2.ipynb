## 불(bool) 인덱싱
-

```python
w= df['max_wind'] >= 10  #True인 값만 보여줌
df[w]
```

```python
df[df['max_wind'] >= 10]
```

조건에 맞는 데이터 추출

```python
# 최고로 바람이 강했던 날  정보(행)
w = df['max_wind'] == df['max_wind'].max()
df[w]
```

## 조건이 2개 이상 --> 그리고 &, 또는 |

```python
## 기온이 30도 이상이고, 최대 풍속이 9이상인 데이터

w =(df['temp'] >= 30) | (df['max_wind'] >= 9)  # or, 또는
df[w]
```

```python
## 기온이 30도 이상이고, 최대 풍속이 9이상인 데이터

w =(df['temp'] >= 30) & (df['max_wind'] >= 9) # and, 그리고
df[w]
```
